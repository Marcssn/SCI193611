```python

P_S = {"true": 0.30, "false": 0.70}

P_C_given_S = {
    "true": {"true": 0.05, "false": 0.95},  # P(C | S = true)
    "false": {"true": 0.01, "false": 0.99},  # P(C | S = false)
}

P_T_given_C = {
    "true": {"บวก": 0.90, "ลบ": 0.10},  # P(T | C = true)
    "false": {"บวก": 0.20, "ลบ": 0.80},  # P(T | C = false)
}


def joint_prob(s, c, t):
    return P_S[s] * P_C_given_S[s][c] * P_T_given_C[c][t]



ans_41 = joint_prob("true", "true", "บวก")
print(f"ข้อ 41: P(S=true, C=true, T='บวก') = {ans_41:.4f}")
print(f"ตอบ b) 0.0135\n")


P_T_pos = sum(
    joint_prob(s, c, "บวก")
    for s in ["true", "false"]
    for c in ["true", "false"]
)
print(f"ข้อ 42: P(T='บวก') = {P_T_pos:.4f}")
print(f"ตอบ b) 0.2154\n")


P_C_true_T_pos = sum(joint_prob(s, "true", "บวก") for s in ["true", "false"])
ans_43 = P_C_true_T_pos / P_T_pos
print(f"ข้อ 43: P(C=true | T='บวก') = {ans_43:.4f} (ประมาณ {ans_43:.3f})")
print(f"ตอบ b) 0.092\n")


P_S_true_T_pos = sum(joint_prob("true", c, "บวก") for c in ["true", "false"])
ans_44 = P_S_true_T_pos / P_T_pos
print(f"ข้อ 44: P(S=true | T='บวก') = {ans_44:.4f} (ประมาณ {ans_44:.3f})")
print(f"ตอบ a) 0.327\n")


N = 1000
expected_cases = N * P_C_true_T_pos
print(f"ข้อ 45: จำนวนผู้ป่วย = 1,000 * P(C=true, T='บวก')")
print(f"= 1,000 * {P_C_true_T_pos:.4f} = {expected_cases:.2f} ราย")
print(f"ตอบ b) ≈ 20 (คำนวณได้ 19.8 ≈ 20 ราย)")
```
